# Exploration (thin)
Calls `src/` only; the pipeline of record is `run_all.py`.

```python
import sys, os; sys.path.insert(0, os.path.abspath('..'))
import numpy as np, pandas as pd
from src.data import load_matches, CORE_SEASONS
from src.model import make_design, fit_poisson, rates, outcome_probs
df = load_matches(CORE_SEASONS)
df.groupby('season')['outcome'].value_counts(normalize=True).unstack().round(3)
```

```python
d = make_design(df[df.season == '2023-24'])
f = fit_poisson(d)
print('gamma =', round(f.gamma, 3), ' free params =', f.n_params)
pd.DataFrame({'team': f.teams, 'alpha': f.alpha, 'beta': f.beta}).sort_values('alpha', ascending=False)
```

```python
lam, mu = rates(f.log_alpha, f.log_beta, f.log_gamma, d.home, d.away)
outcome_probs(lam, mu).mean(axis=0)  # mean P(H), P(D), P(A)
```
